```python
def calcular_media(numeros):
  if not numeros:
    return 0  # Return 0 for an empty list or handle as an error
  return sum(numeros) / len(numeros)

# Exemplo de uso:
lista_de_numeros = [10, 20, 30, 40, 50]
media = calcular_media(lista_de_numeros)
print(f"A média dos números {lista_de_numeros} é: {media}")

lista_vazia = []
media_vazia = calcular_media(lista_vazia)
print(f"A média da lista vazia {lista_vazia} é: {media_vazia}")
```
